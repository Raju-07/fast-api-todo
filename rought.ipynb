```python
from sqlalchemy import func
from datetime import date,timedelta
today = date.today()
expired_at = today + timedelta(days=23)

time_diff = func.date_part('day',expired_at-today)
print(time_diff)
```

```python
from sqlalchemy import func, cast, Date

day_diff = func.julianday(expired_at) - func.julianday(today)
print(day_diff)
```
